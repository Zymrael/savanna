```python
from datatrove.pipeline.readers import ParquetReader
import json
import os
```

```python
DATA_PATH = ""
```

```python
def write_jsonl(documents, output_file):
    """Write documents to a JSONL file."""
    os.makedirs(os.path.dirname(output_file), exist_ok=True)
    with open(output_file, 'w', encoding='utf-8') as f:
        for doc in documents:
            json_str = json.dumps({'text': doc.text})
            f.write(json_str + '\n')

# Create output directory if it doesn't exist
output_file = f"{DATA_PATH}/fineweb_slice_0.jsonl"

# Read documents and write to JSONL
data_reader = ParquetReader("hf://datasets/HuggingFaceFW/fineweb/data", limit=300000)
write_jsonl(data_reader(), output_file)
```
